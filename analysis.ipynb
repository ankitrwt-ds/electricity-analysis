```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("data/dataset_tk.csv")
df
```

```python
# Change Unnamed: 0 to Date

df.rename(columns={"Unnamed: 0": "Date"}, inplace=True)
df["Date"] = pd.to_datetime(df["Date"], dayfirst=True)
df.sort_values("Date", inplace=True)
df.reset_index(drop=True, inplace=True)
```

```python
df.columns
df.head()
```

```python
df.info()
```

```python
df.shape
```

```python
# Data Cleaning and Understanding

df.isnull().sum() # Check missing value
df.duplicated().sum() # Check duplicates
df.dtypes # Check data type
df.describe() # Statistical summary
```

```python
# Date Based Features

df["Year"] = df["Date"].dt.year
df["Month"] = df["Date"].dt.month
df["Month_Name"] = df["Date"].dt.month_name()
df["Day"] = df["Date"].dt.day
df["Day_Name"] = df["Date"].dt.day_name()

df.head()
```

```python
# Total Power Consumption All Over the country per day

state_col = df.columns.difference(["Date","Year","Month","Month_Name","Day","Day_Name"])
df["Total_Consumption"] = df[state_col].sum(axis=1)
df.head()
```

```python
# Total Consumption by State

total_by_state = round(df[state_col].sum().sort_values(ascending=False),2)
total_by_state.head()
```

```python
# Highest & Lowest Consumption State

highest_state = total_by_state.idxmax()
highest_consumption = total_by_state.max()

print("Highest Consuming State:", highest_state)
print("Total Consumption:", highest_consumption, "MU")

lowest_state = total_by_state.idxmin()
lowest_consumption = total_by_state.min()

print("Lowest Consuming State:", lowest_state)
print("Total Consumption:", lowest_consumption, "MU")
```

```python
# Daily Power Consumption

# 1. Average daily consumption

daily_average = round(df["Total_Consumption"].mean(),2)

print("Average Daily Power Consumption:", daily_average, "MU")

# 2. Highest consumption day

max_day = df.loc[df["Total_Consumption"].idxmax()]

print("Highest Consumption Date:", max_day["Date"])
print("Highest Consumption:", max_day["Total_Consumption"], "MU")

# 3. Lowest consumption day

min_day = df.loc[df["Total_Consumption"].idxmin()]

print("Lowest Consumption Date:", min_day["Date"])
print("Lowest Consumption:", min_day["Total_Consumption"], "MU")
```

```python
# Month-wise Power Consumption

month_avg =round(df.groupby(["Year","Month"])["Total_Consumption"].mean().reset_index(),2)
print(month_avg.head(3))

# Monthly Total Consumption

monthly_total = df.groupby(
    ["Year", "Month"]
)["Total_Consumption"].sum().reset_index()
print(monthly_total.head(3))
```

```python
# State-wise Average Consumption

avg_by_state = round(df[state_col].mean().sort_values(ascending = False),2)
print(avg_by_state.head(3))

# State-wise Maximum Consumption

max_by_state = round(df[state_col].max().sort_values(ascending = False),2)
print(max_by_state.head(3))

# State-wise Minimum Consumption

min_by_state = round(df[state_col].min().sort_values(ascending = True),2)
print(min_by_state.head(3))
```

```python
# Visualization 1 : Top 10 State by Total Power Consumption

top10_state = total_by_state.head(10)

# plt.figure(figsize=(10, 6))
top10_state.plot(kind = "bar")

plt.title("Top 10 State by Total Power Consumption")
plt.xlabel("State")
plt.ylabel("Total Power Consumption (MU)")
plt.xticks(rotation = 75)

plt.savefig("visualizations/top10_states.png", dpi=300, bbox_inches="tight")

plt.show()
```

### Outcome

The bar chart shows the top 10 states with the highest total power consumption during the analyzed period. The state with the tallest bar has the highest overall electricity consumption, while the state with the shortest bar among the top 10 has the lowest consumption within this group. The consumption values are measured in Mega Units (MU).

```python
# Visualization 2 : Year-wise Power Consumption

yearly_consumption = df.groupby("Year")["Total_Consumption"].sum()

print(yearly_consumption)

yearly_consumption.plot(kind="bar")

plt.title("Year-wise Total Power Consumption")
plt.xlabel("Year")
plt.ylabel("Total Power Consumption (MU)")
plt.xticks(rotation=0)

plt.show()
```

### Outcome

The bar chart compares the total power consumption across different years in the dataset. It clearly shows the variation in annual electricity consumption and helps identify the year with the highest and lowest total consumption.

```python
# Visualization 3 : Monthly Consumption Visualization

monthly_total["Month_Date"] = pd.to_datetime(
    monthly_total["Year"].astype(str) + "-" +
    monthly_total["Month"].astype(str) + "-01"
)


plt.plot(
    monthly_total["Month_Date"],
    monthly_total["Total_Consumption"]
)

plt.title("Monthly Total Power Consumption")
plt.xlabel("Month")
plt.ylabel("Total Power Consumption (MU)")
plt.xticks(rotation=45)

plt.savefig("visualizations/monthly_total.png", dpi=300, bbox_inches="tight")

plt.show()
```

### Outcome

The line chart shows the monthly total power consumption from 2019 to 2020. The consumption was generally high during 2019, with the highest value occurring around mid-2019. A sharp decline can be observed in early 2020, after which the monthly consumption remained at a much lower level.

```python
# Visualization 4 : Top 10 Average Consumption by State

top10_avg_states = avg_by_state.head(10)

top10_avg_states.plot(kind="bar")

plt.title("Top 10 States by Average Daily Power Consumption")
plt.xlabel("State")
plt.ylabel("Average Daily Power Consumption (MU)")
plt.xticks(rotation=75)

plt.show()
```

### Outcome

The bar chart shows the top 10 states based on their average daily power consumption during the analyzed period. The visualization makes it easy to compare the typical daily electricity consumption levels of the states. States with taller bars have higher average daily consumption.

The bar chart compares the average daily power consumption for each month across the analyzed years. It helps identify the months with relatively higher and lower typical electricity consumption and shows the overall seasonal pattern in power usage.

```python
# Visualization 6 : Top 10 states by their highest single-day power consumption.

top10_max_state = max_by_state.head(10)

top10_max_state.plot(kind="bar")

plt.title("Top 10 States by Maximum Daily Power Consumption")
plt.xlabel("State")
plt.ylabel("Maximum Daily Power Consumption (MU)")
plt.xticks(rotation=75)

plt.show()
```

### Outcome

The bar chart shows the top 10 states with the highest recorded single-day power consumption during the analyzed period. It allows comparison of the peak daily electricity demand among different states. States with taller bars experienced higher maximum daily consumption.

```python
# Visualization 7 : 10 states by their lowest single-day power consumption.

bottom10_min_state = min_by_state.head(10)

bottom10_min_state.plot(kind="bar")

plt.title("10 States with Lowest Minimum Daily Power Consumption")
plt.xlabel("State")
plt.ylabel("Minimum Daily Power Consumption (MU)")
plt.xticks(rotation=75)

plt.show()
```

### Outcome

The bar chart shows the 10 states with the lowest recorded single-day power consumption during the analyzed period. It provides a comparison of the minimum daily consumption levels among the states and highlights states with relatively lower electricity demand.

```python
# State-wise Correlation Analysis

correlation_matrix = df[state_col].corr()
print(correlation_matrix.head(3))

# Clean Top Correlated State Pairs

corr_pairs = correlation_matrix.where(
    np.triu(np.ones(correlation_matrix.shape), k=1).astype(bool)
)

corr_pairs = corr_pairs.stack().sort_values(ascending=False)

corr_pairs.head(10)
```

```python
# Visualization 8 : Top 10 State Correlations Visualization

top10_corr = corr_pairs.head(10).sort_values(ascending=True)

top10_corr.plot(kind="barh")

plt.title("Top 10 Strongest Correlations Between States")
plt.xlabel("Correlation Coefficient")
plt.ylabel("State Pair")

plt.savefig("visualizations/correlation.png", dpi=300, bbox_inches="tight")

plt.show()
```

### Outcome

The correlation chart shows the 10 strongest positive relationships between state-wise power consumption. Haryana and Punjab have the strongest correlation, with a coefficient of approximately 0.956. The high correlation values indicate that these state pairs tend to show similar patterns in their power consumption over time.

```python
# Calculate yearly consumption for each state

state_year_consumption = df.groupby("Year")[state_col].sum()

print(state_year_consumption)

# Visualization : 9 Top 10 states by total consumption across the available years

top10_states = state_year_consumption.sum().sort_values(
    ascending=False
).head(10).index

state_year_top10 = state_year_consumption[top10_states]

state_year_top10.plot(
    kind="bar"
)

plt.title("Year-wise Power Consumption of Top 10 States")
plt.xlabel("Year")
plt.ylabel("Total Power Consumption (MU)")
plt.xticks(rotation=0)

plt.legend(title="State")
plt.tight_layout()

plt.savefig("visualizations/yearly_top10.png", dpi=300, bbox_inches="tight")

plt.show()
```

### Outcome

The visualization compares the total power consumption of the top 10 states for 2019 and the available period of 2020. Maharashtra recorded the highest consumption among the selected states, followed by Gujarat and UP. The ranking of the major consuming states can be clearly observed from the chart. The 2020 values are lower because the dataset contains only part of 2020, so they should not be interpreted as complete-year consumption.
